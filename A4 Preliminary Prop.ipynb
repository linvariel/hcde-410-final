## Motivation/problem statement

**Why are you planning to do this analysis?**

I plan to analyze wildlife-vehicle collisions over time in Banff National Park because the park has implemented many wildlife crossing structures since its founding.

**Why is it potentially interesting and useful, from a scientific, practical, and/or human-centered perspective?**

Wildlife-vehicle collisions are a costly problem that seems to be normalized or considered inevitable. According to a [Pew Research report](https://www.pew.org/en/research-and-analysis/articles/2021/05/10/wildlife-vehicle-collisions-are-a-big-and-costly-problem-and-congress-can-help), "An estimated 1 million to 2 million crashes between motor vehicles and large animals such as deer occur every year in the U.S., causing approximately 200 human deaths, 26,000 injuries, and at least $8 billion in property damage and other costs." National parks today are tasked with ecological conservation and human safety. Wildlife crossing structures like over and underpasses have the potential to reduce costly and sometimes fatal vehicle collisions with wildlife, protecting humans, animals, and saving money.

**What do you hope to learn? (Overall research goal, specific hypotheses or research questions aren’t necessary at this point)**

I hope to learn about the overall usage of wildlife crossing structures at Banff National Park, such as the types of animals, which structures are most often used, and whether the time of year affects usage. I would also like to investigate the number of animal deaths caused by vehicle collisions.

## Data used

**What dataset do you plan to use, and why?**

I plan to use the Banff Wildlife Crossing Structures and Banff Wildlife Mortality datasets from Canada's open data website.

**Summarize what is represented in the dataset**

The crossing structures dataset includes the dates, locations, and counts of different wildlife crossing structure users, including humans. The time range of this data is only a few years, from 2014 to 2017.

The mortality dataset includes the dates, types of wildlife, and causes of death. This dataset has a longer time range and many more entries than the other dataset.

**Link to the dataset, and specify license/terms of use**

[Wildlife Mortality - Banff](https://open.canada.ca/data/en/dataset/d5fd45b9-7c70-4339-a68b-e60d7cbf42e8)
[Wildlife Crossing Structures - Banff](https://open.canada.ca/data/en/dataset/6aa18934-5fec-4f12-8dd7-8356555d0576)

Both datasets use the [Open Government Licence - Canada](https://open.canada.ca/en/open-government-licence-canada)

**Briefly justify why this dataset is relevant to your problem statement**

These datasets are relevant to my exploration of usage and effects of wildlife crossing structures: the first dataset includes the different kinds of users, and the second dataset can provide insights over time of wildlife mortalities caused by road-crossing.

**Are there ethical considerations to using this dataset? What and why, or why not?**

Neither of the datasets includes sensistive identifiable information, so there are not significant ethical considerations. The crossing structures dataset includes counts, which includes counting human users, but these users are not in any way identified. However, human users might not be aware that they are being counted when the cross the structures.
